```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
```

```python
df['Order_Date'].dtype
```

```python
monthly_sales = df.groupby(
    df['Order_Date'].dt.to_period('M')
)['Sales_Amount'].sum()
```

```python
 monthly_sales_table = monthly_sales.reset_index()
 monthly_sales_table.columns = ['Month', 'Total Sales']
 monthly_sales_table['Total Sales'] = monthly_sales_table['Total Sales'].round(2)
 monthly_sales_table.style.format({
    'Total Sales': '{:,.2f}'
})

 monthly_sales_table
```

```python
plt.figure(figsize=(15, 6))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker='o'
)

plt.title('Monthly Sales Trend (2022–2025)')
plt.xlabel('Month')
plt.ylabel('Total Sales Amount')

plt.xticks(
    range(0, len(monthly_sales), 3),
    monthly_sales.index.astype(str)[::3],
    rotation=45
)

plt.grid(True)
plt.tight_layout()
plt.show()
```

### Monthly Sales Trend – Initial Observations

1. Monthly sales fluctuate throughout the period from 2022 to 2025.
2. December 2025 recorded the highest monthly sales of 270,672.91.
3. January 2022 recorded the lowest monthly sales of 95,723.86.
4. Sales were relatively higher during October, November and December in several years.
5. Overall, the monthly sales data indicates variations in sales performance across different months and years.

```python
quarterly_sales = df.groupby(
    df['Order_Date'].dt.to_period('Q')
)['Sales_Amount'].sum()
```

```python
quarterly_sales_table = quarterly_sales.reset_index()
quarterly_sales_table.columns = ['Quarter', 'Total Sales']
quarterly_sales_table['Total Sales'] = quarterly_sales_table['Total Sales'].round(2)
quarterly_sales_table.style.format({
    'Total Sales': '{:,.2f}'
})

quarterly_sales_table
```

```python
plt.figure(figsize=(13, 6))

plt.plot(
    quarterly_sales.index.astype(str),
    quarterly_sales.values,
    marker='o'
)

plt.title('Quarterly Sales Trend (2022–2025)')
plt.xlabel('Quarter')
plt.ylabel('Total Sales Amount')

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

### Quarterly Sales Trend – Key Observations

1. Quarterly sales fluctuated across the four-year period from 2022 to 2025.
2. The fourth quarter recorded the highest sales in each year.
3. The highest quarterly sales were recorded in Q4 2025, amounting to 758,331.65.
4. The lowest quarterly sales were recorded in Q1 2024, amounting to 404,729.76.
5. Overall, Q4 consistently demonstrated stronger sales performance compared to other quarters.

```python
age_bins = [0, 17, 25, 35, 45, 55, np.inf]

age_labels = [
    'Below 18',
    '18-25',
    '26-35',
    '36-45',
    '46-55',
    'Above 55'
]

df['Age_Group'] = pd.cut(
    df['Customer_Age'],
    bins=age_bins,
    labels=age_labels
)

age_group_table = (
    df['Age_Group']
    .value_counts()
    .sort_index()
    .reset_index()
)

age_group_table.columns = [
    'Age Group',
    'Number of Transactions'
]

age_group_table
```

```python
age_counts = df['Age_Group'].value_counts().sort_index()

plt.figure(figsize=(10, 5))

age_counts.plot(kind='bar')

plt.title('Customer Age Group Distribution')
plt.xlabel('Age Group')
plt.ylabel('Number of Transactions')

plt.xticks(rotation=0)
plt.grid(axis='y')
plt.tight_layout()

plt.show()
```

### Customer Age Group Analysis – Key Observations

1. The 36–45 age group recorded the highest number of transactions, with 6,279 records.
2. The 26–35 age group was the second-largest segment, with 4,812 records.
3. The below-18 age group recorded only 2 transactions.
4. Customers aged 26–45 account for a substantial share of the transaction records.
5. The age group above 55 recorded 1,164 transactions.
6. The business may focus on understanding the purchasing behaviour of customers aged 26–45.

```python
df['Category_Clean'] = (
    df['Product_Category']
    .str.strip()
    .str.title()
)

df['Category_Clean'].value_counts(dropna=False)
```

```python
category_revenue = (
    df.groupby('Category_Clean')['Sales_Amount']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

category_revenue.columns = [
    'Product Category',
    'Total Revenue'
]

category_revenue
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    category_revenue['Product Category'],
    category_revenue['Total Revenue']
)

plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

### Revenue by Product Category – Key Observations

1. Electronics generated the highest revenue of 4,932,145.47.
2. Furniture ranked second with revenue of 1,989,859.24.
3. Office Supplies generated revenue of 745,040.02.
4. Appliances recorded the lowest revenue of 585,096.67.
5. Electronics contributed approximately 59.8% of total revenue.
6. The business may focus on maintaining inventory availability and reviewing sales strategies for high-revenue categories.

```python
correlation_matrix = df[numerical_columns].corr()

correlation_matrix
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Variables')
plt.tight_layout()

plt.show()
```

### Correlation Heatmap – Key Observations

1. Sales Amount and Profit show a strong positive correlation of 0.9174.
2. Sales Amount and Cost Amount have a very strong positive correlation of 0.9955.
3. Unit Price has a moderate positive correlation with Sales Amount (0.6841).
4. Quantity has a moderate positive correlation with Sales Amount (0.4283).
5. Discount Percentage has a weak negative correlation with Profit (-0.1911).
6. Customer Age has almost no linear correlation with Sales Amount.
7. Correlation indicates association between variables and does not establish causation.

# Conclusion

The Exploratory Data Analysis of Retail Sales Data covering 2022–2025 provided useful insights into sales performance, customer behaviour, product demand and business profitability.

The dataset contains 18,045 transaction records and 36 columns. The analysis identified missing values, duplicate records and unusual numerical values that require further verification.

The sales analysis showed fluctuations in monthly and quarterly performance, with the highest monthly sales of 270,672.91 recorded in December 2025 and the highest quarterly sales of 758,331.65 recorded in Q4 2025.

Customer analysis revealed that the 36–45 age group contributed the highest number of transaction records. Female transactions were slightly higher than male transactions.

Product analysis identified Copy Paper Box as the top-selling product by quantity, while Electronics generated the highest category revenue of 4,932,145.47.

The correlation analysis showed a strong positive relationship between Sales Amount and Profit. The sales distribution histogram also indicated a positively skewed distribution, with most transactions concentrated in lower sales-value ranges.

Overall, this analysis provides a better understanding of the retail business and highlights opportunities to improve inventory planning, sales strategies and profitability.

# Business Recommendations

1. **Focus on High-Revenue Categories:** Electronics generates the highest revenue. The business should maintain sufficient inventory of high-performing electronic products and monitor their sales regularly.

2. **Improve Inventory Planning:** Products such as Copy Paper Box and Precision Mouse have high sales volumes. Inventory levels should be reviewed regularly to reduce stock shortages and improve product availability.

3. **Strengthen Customer Segmentation:** Customers aged 26–45 represent a substantial portion of transaction records. The business can study their purchasing preferences and develop targeted marketing campaigns.

4. **Review Discount Strategies:** The analysis indicates a weak negative correlation between Discount Percentage and Profit. The business should evaluate discount campaigns carefully to protect profit margins.

5. **Investigate Data Quality Issues:** Duplicate records, missing values, negative quantities and unusual discount values should be reviewed to improve the reliability of future business reports.
